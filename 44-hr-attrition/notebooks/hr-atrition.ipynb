# 📂 Overview

* **Background** <br>
  Employee attrition is one of the most critical challenges organizations face in maintaining workforce stability and operational efficiency. High employee turnover can lead to increased recruitment costs, productivity loss, reduced team morale, and disruptions in business continuity.

  Understanding the key drivers behind employee resignation is essential for organizations to improve employee satisfaction, strengthen retention strategies, and build a more sustainable work environment.

* **Goal of the Project** <br>
  Using historical employee demographic, compensation, workplace satisfaction, and organizational data, this project aims to build predictive models that identify employees who are more likely to leave the company.

  By leveraging machine learning techniques and exploratory data analysis (EDA), organizations can proactively detect attrition risks and make data-driven HR decisions to improve workforce retention and organizational performance.

🎯 Key Features

**Target Variable**

* `Attrition`: Employee attrition status (`Stayed` / `Left`)

**Employee Demographics**

* `Age`: Employee age
* `Gender`: Gender
* `Marital Status`: Marital status
* `Education Level`: Highest education qualification
* `Number of Dependents`: Number of dependents

**Career & Employment Information**

* `Job Role`: Employee job function or department
* `Job Level`: Employee seniority level
* `Years at Company`: Total years employed at the company
* `Company Tenure`: Overall tenure within the organization
* `Number of Promotions`: Number of promotions received

**Compensation & Benefits**

* `Monthly Income`: Employee monthly salary

**Work Environment & Satisfaction**

* `Work-Life Balance`: Employee perception of work-life balance
* `Job Satisfaction`: Employee job satisfaction level
* `Performance Rating`: Performance evaluation score
* `Employee Recognition`: Recognition level received by employee
* `Company Reputation`: Employee perception of company reputation

**Work Arrangement & Accessibility**

* `Remote Work`: Remote or onsite working status
* `Distance from Home`: Distance between employee residence and workplace
* `Overtime`: Overtime working status

**Organizational Growth Opportunities**

* `Leadership Opportunities`: Availability of leadership opportunities
* `Innovation Opportunities`: Availability of innovation and growth opportunities

**Company Information**

* `Company Size`: Organization size category

📁 Files Provided

* `train.csv`: Main dataset used for exploratory data analysis and machine learning modeling
* `test.csv`: Main dataset used for testing.

🚀 Project Objective

The primary objective of this project is to perform comprehensive exploratory data analysis (EDA) and develop machine learning models to predict employee attrition behavior.

The project focuses on identifying the key factors influencing employee turnover and building predictive systems that help organizations proactively manage retention strategies.

By leveraging these insights, organizations can:

* Improve **employee retention strategies**
* Reduce **turnover-related operational costs**
* Enhance **employee satisfaction and engagement**
* Support **data-driven HR decision-making**
* Identify **high-risk attrition groups early**

**AUC-ROC Overview**

**AUC-ROC** evaluates a model’s ability to **distinguish between two classes**.

* **ROC curve** plots **TPR vs FPR**
* **AUC** = Area under ROC curve → **ranking quality of predictions**

**Why AUC-ROC?**

* Robust to **class imbalance**
* Measures **ranking ability**, not fixed threshold (0.5)
* Common metric in **credit risk & fraud prediction**

> Higher AUC → better separation between *Paid* vs *Default*

**Interpretation**

| AUC Score     | Meaning                  |
| ------------- | ------------------------ |
| **1.0**       | Perfect model            |
| **0.9+**      | Excellent discrimination |
| **0.7 – 0.9** | Good                     |
| **0.5**       | Random guessing          |
| **< 0.5**     | Worse than random        |

(Source: [ROC AUC](https://en.wikipedia.org/wiki/Receiver_operating_characteristic))

```python
logic_auc = roc_auc_score(y_train, oof_pred_logic)
ridge_auc = roc_auc_score(y_train, oof_pred_ridge)
linearsvc_auc = roc_auc_score(y_train, oof_pred_linearsvc)

# AUC-weighted blend
total_auc = logic_auc + ridge_auc + linearsvc_auc
w_logic = logic_auc / total_auc
w_ridge = ridge_auc / total_auc
w_linearsvc = linearsvc_auc / total_auc

blended_oof = w_logic * oof_pred_logic + w_ridge * oof_pred_ridge + w_linearsvc * oof_pred_linearsvc
blended_test = w_logic * test_pred_logic + w_ridge * test_pred_ridge + w_linearsvc * test_pred_linearsvc

logger.info("*"*80)
logger.info(f"LogisticRegression OOF AUC: {logic_auc:.5f} (Weight: {w_logic:.6f})")
logger.info(f"RidgeClassifier OOF AUC: {ridge_auc:.5f} (Weight: {w_ridge:.6f})")
logger.info(f"LinearSVC OOF AUC: {linearsvc_auc:.5f} (Weight: {w_linearsvc:.6f})")
logger.info(f"Blended OOF AUC: {roc_auc_score(y_train, blended_oof):.6f}")
logger.info("*"*80)

fig, ax = plt.subplots(1, 4, figsize=(20, 5))

ax[0].hist(test_pred_logic, bins=50, color=CFG.COLORS["primary"], alpha=0.7, edgecolor="white")
ax[0].set_title("LogisticRegression OOF Predictions", fontsize=12, fontweight="bold", pad=15, color=CFG.COLORS["dark"])
ax[0].set_xlabel("Predicted Probability", fontsize=10)
ax[0].grid(axis="y", alpha=0.3)

ax[1].hist(test_pred_ridge, bins=50, color=CFG.COLORS["accent"], alpha=0.7, edgecolor="white")
ax[1].set_title("RidgeClassifier OOF Predictions", fontsize=12, fontweight="bold", pad=15, color=CFG.COLORS["dark"])
ax[1].set_xlabel("Predicted Probability", fontsize=10)
ax[1].grid(axis="y", alpha=0.3)

ax[2].hist(test_pred_linearsvc, bins=50, color=CFG.COLORS["secondary"], alpha=0.7, edgecolor="white")
ax[2].set_title("LinearSVC OOF Predictions", fontsize=12, fontweight="bold", pad=15, color=CFG.COLORS["dark"])
ax[2].set_xlabel("Predicted Probability", fontsize=10)
ax[2].grid(axis="y", alpha=0.3)

ax[3].hist(blended_test, bins=50, color=CFG.COLORS["highlight"], alpha=0.7, edgecolor="white")
ax[3].set_title("Blended Predictions", fontsize=12, fontweight="bold", pad=15, color=CFG.COLORS["dark"])
ax[3].set_xlabel("Predicted Probability", fontsize=10)
ax[3].grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
